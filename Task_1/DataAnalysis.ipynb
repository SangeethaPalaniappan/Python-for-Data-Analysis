```python
import pandas as pd
import warnings 
warnings.filterwarnings('ignore')
```

```python
# Reading data from the csv files

sales_trans_data_2017 = pd.read_csv('Sales_Trans_2017.csv')
sales_trans_data_2018 = pd.read_csv('Sales_Trans_2018.csv')
sales_trans_data_2019 = pd.read_csv('Sales_Trans_2019.csv')
```

```python
# Concatenate the csv files

concated_sales_trans_data = pd.concat([sales_trans_data_2017,sales_trans_data_2018,sales_trans_data_2019])
concated_sales_trans_data
```

```python
# To know the shape(rows, columns)

concated_sales_trans_data.shape
```

```python
# To check how many null values in each column

concated_sales_trans_data.isnull().sum()
```

```python
# Delete the rows

del concated_sales_trans_data['Gross']
del concated_sales_trans_data['Disc']
del concated_sales_trans_data['Voucher Amount']
```

```python
concated_sales_trans_data
```

```python
# Drop the rows where all the values in the row is null

concated_sales_trans_data.dropna(axis=0, how='all', inplace=True)
concated_sales_trans_data
```

```python
# Drop the row which has the null values greater than the thresh value

concated_sales_trans_data.dropna(thresh=4, inplace=True)
```

```python
concated_sales_trans_data.shape
```

```python
# To get the rows which has the Product value empty

concated_sales_trans_data[concated_sales_trans_data['Product'].str.strip()=='']
```

```python
# Keep the product column only has values by removing empty values

concated_sales_trans_data = concated_sales_trans_data[concated_sales_trans_data['Product'].str.strip()!='']
```

```python
# Replace the value 

voucher = concated_sales_trans_data['Voucher'].copy()
concated_sales_trans_data['Voucher']=concated_sales_trans_data['Voucher'].str.replace("Sal:",'')
voucher
```

```python
# Convert the Voucher column into integer

concated_sales_trans_data['Voucher']=concated_sales_trans_data['Voucher'].astype(int)
```

```python
# Convert the Qty column into integer
qty = concated_sales_trans_data['Qty'].copy()
concated_sales_trans_data['Qty']=pd.to_numeric(concated_sales_trans_data['Qty'].str.replace(",",""),errors='coerce').astype(int)
qty
```

```python
# Convert the Rate column into float
rate = concated_sales_trans_data['Rate'].copy()
concated_sales_trans_data['Rate']=pd.to_numeric(concated_sales_trans_data['Rate'].str.replace(",",""), errors='coerce').astype('float')
rate
```

```python
# Convert the Date column into datetime
date = concated_sales_trans_data['Date']
concated_sales_trans_data['Date']=pd.to_datetime(concated_sales_trans_data['Date'], format='%d/%m/%Y')
date
```
