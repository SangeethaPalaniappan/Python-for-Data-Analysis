## Who uses Social media more - Male or Female?

```python
sns.countplot(data=sm_impact_data, x='Gender', color='purple')
plt.ylabel('No. of Users')
plt.title('Usage og Social Media based on Gender')
plt.savefig('More usage.png')
plt.show()
```

### From the above data Female uses the most comparing others

## Which age group has the highest average daily social media usage?

```python
sns.countplot(data=sm_impact_data, x='Age', color='green')
plt.ylabel('No. of Users')
plt.savefig('Age Group.png')
```

## Which Platform is used more?

```python
sns.countplot(data=sm_impact_data, x='Primary_Platform', width=0.7, color='purple')
plt.ylabel('No. of Users')
plt.savefig('Platforms.png')
```

```python
platforms = sm_impact_data.Primary_Platform.value_counts()
plt.pie(platforms, labels= ['Instagram', 'TikTok', 'YouTube', 'Snapchat', 'Reddit', 'X (Twitter)', 'LinkedIn'], autopct='%1.0f%%', shadow=True)
plt.title('Platforms Used')
plt.savefig('Platforms.png')
plt.show()
```

### Instagram tops the most usage Platform

## Which Device type?

```python
sns.countplot(data=sm_impact_data, x='Device_Type', color='orange')
plt.ylabel('No. of Users')
plt.title('Device Type')
plt.savefig('Device_type.png')
plt.show()
```

### Smart Phones used mostly from the above data

## Social Comparison 

```python
sns.countplot(data=sm_impact_data,x='Social_Comparison_Frequency', color= 'maroon')
plt.ylabel('No. of Users')
plt.title('Social Comparison rate')
plt.savefig('Social Comparison rate')
plt.show()
```

```python
frequency = sm_impact_data.Social_Comparison_Frequency.value_counts()
frequency
plt.pie(frequency, labels = ['Sometimes','Frequently','Rarely','Always','Never'], autopct='%1.0f%%')
plt.title('Social Comparison Frequency')
plt.savefig('Social Comparison Frequency')
plt.show()
```

## Does Over usage of Phone Affects the sleep quality?

```python
sns.barplot(data=sm_impact_data, x='Sleep_Quality_Score', y='Daily_Usage_Hours', errorbar=None)
plt.title('Relationship Between Sleep Quality and Daily Usage')
plt.savefig('Sleep Quality.png')
plt.show()
```

### Higher social-media usage is associated with lower sleep quality

## Sleep Duration hours affects the Mental health index?

```python
sns.scatterplot(data=sm_impact_data,y='Sleep_Duration_Hours', x='Mental_Health_Index')
plt.title('Relationship Between Mental Health Index and Sleep Duration')
plt.savefig('Mental Health Index and Sleep Duration.png')
plt.show()
```

## Daily_Usage_Hours affects the CGPA?

```python
sns.scatterplot(data=sm_impact_data,y='Daily_Usage_Hours', x='Academic_Performance_GPA')
plt.title('Daily_Usage_Hours vs Academic_Performance_GPA')
plt.savefig('Usage vs Performance.png')
plt.show()
```
